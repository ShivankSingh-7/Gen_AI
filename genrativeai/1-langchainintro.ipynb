### Streaming And Batch

### Streaming
Streaming = data comes piece-by-piece in real time instead of all at once.


Most models can stream their output content while it is being generated. By displaying output progressively, Streaming significantly improves user experience, particularly for longer responses. Calling Streanm() returns an iterator that yeild output chunks as they are produced.we can use a loop to process them in real time 

```python
for chunk in llm.stream("write a 1000 word para on Modi"):
    print(chunk.text, end="", flush=True)
```

### Batch
process everything together then return the full result at once

all the input goes at once then processing occurs parallely and then return the response at once 

```python
responses = llm.batch(
    [
        "why do parots have colorful feather",
        "why human can not fly",
        "why elon musk is so rich"
    ]
)
```

```python
for response in responses:
    print(response.content)
```
