### Generative AI Application With Groq AI Model

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0,
    max_retries=2    
    
)
```

```python
llm
```

```python
response = llm.invoke("write me an essay on AI")
response
```

```python
print(response.content)
```

```python
### using langchain.chat_models
from langchain.chat_models import init_chat_model
```

```python
llm = init_chat_model(
    "llama-3.3-70b-versatile",
    model_provider="groq"
)
```

```python
response = llm.invoke("write an essay about donald trump attacking on iran")
print(response.content)
```
